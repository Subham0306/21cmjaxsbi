```python
fX_list = []
tau_list = []
p3_list = []
logMmin_list = []
fesc_list = []

for s in sorted(os.listdir(".")):
    if s.startswith("simu"):
        param_path = os.path.join(s, "parameters.txt")

        if not os.path.exists(param_path):
            continue

        with open(param_path, "r") as f:
            lines = [line.strip() for line in f if line.strip()]

        if len(lines) < 5:
            continue   # skip empty/incomplete simulations

        values = [float(line.split()[0]) for line in lines[:5]]

        fX_list.append(values[0])
        tau_list.append(values[1])
        p3_list.append(values[2])
        logMmin_list.append(values[3])
        fesc_list.append(values[4])

fX = np.array(fX_list)
tau = np.array(tau_list)
p3 = np.array(p3_list)
logMmin = np.array(logMmin_list)
fesc = np.array(fesc_list)
```

```python
import os
import numpy as np


def sample_pk(k, Pk, k_bins):
    return np.interp(k_bins, k, Pk)

X_clean = []
Theta = []

idx_choice = [14, 16, 18, 20, 22, 24, 26, 28]

# Exact k-bin centers

k_bins = np.array([
    0.16,
    0.23,
    0.33,
    0.47,
    0.66,
    0.93,
    1.32,
    1.86
])

# Exact redshifts

z_values = np.array([
    13.18242,
    12.06334,
    11.14162,
    10.36687,
    9.704856,
    9.131420,
    8.629000,
    8.184488
])

for s in sorted(os.listdir(".")):

    if not s.startswith("simu"):
        continue

    simu_path = os.path.join(".", s)

    ps_path = os.path.join(
        simu_path,
        "postprocessing",
        "ps_dtb"
    )

    if not os.path.exists(ps_path):
        continue

    try:

        x_list = []

        # Loop over redshifts

        for idx in idx_choice:

            fname = f"powerspectrum{idx:03d}.dat"

            fullpath = os.path.join(ps_path, fname)

            if not os.path.exists(fullpath):
                raise FileNotFoundError(fullpath)

            k, Pk = np.loadtxt(fullpath, unpack=True)

            # Interpolate onto fixed Loreli-II bins
            Pk_sampled = sample_pk(
                k,
                Pk,
                k_bins
            )

            x_list.append(Pk_sampled)

        # Final clean feature vector
        # Shape = 64

        x = np.concatenate(x_list)

        # Load astrophysical parameters

        with open(os.path.join(simu_path, "parameters.txt")) as f:

            lines = [
                line.strip()
                for line in f
                if line.strip()
            ]

        if len(lines) < 5:
            continue

        vals = [
            float(line.split()[0])
            for line in lines[:5]
        ]

        # Original file order:
        # [fX, tau, rhs, logMmin, fesc]

        theta = np.array([
        vals[0],  # fX
        vals[1],  # tau
        vals[2],  # rhs
        vals[3],  # logMmin
        vals[4],  # fesc
      ])

        X_clean.append(x)
        Theta.append(theta)

    except Exception as e:
        continue

        #print(f"{s} failed: {e}")

# Convert to arrays

X_clean = np.array(X_clean)
Theta = np.array(Theta)

print("------------------------------------------------")
print("LORELI-II CLEAN DATASET READY")
print("------------------------------------------------")

print("Redshifts used:")
print(z_values)

print("\nk bins used:")
print(k_bins)

print("\nX_clean shape:", X_clean.shape)
print("Theta shape:", Theta.shape)

print("\nParameter order:")
print("[fX, tau, rhs, logMmin, fesc]")
```

```python
def preprocess_X(
    X_clean,
    mean=None,
    std=None
):

    X = np.clip(
        X_clean,
        1e-6,
        None
    )

    X = np.log10(X)

    if mean is not None:
        X = (X - mean)/std

    X = np.clip(
        X,
        -8.0,
        8.0
    )

    return X

X_train_tmp = preprocess_X(X_train)

X_mean = X_train_tmp.mean(axis=0)

X_std = X_train_tmp.std(axis=0)

X_std[X_std == 0] = 1.0

X_train_proc = preprocess_X(
    X_train,
    X_mean,
    X_std
)

X_val_proc = preprocess_X(
    X_val,
    X_mean,
    X_std
)

X_test_proc = preprocess_X(
    X_test,
    X_mean,
    X_std
)

print("Train:", X_train_proc.shape)
print("Val:", X_val_proc.shape)
print("Test:", X_test_proc.shape)

print("\nTrain feature mean (first 5):")
print(X_train_proc.mean(axis=0)[:5])

print("\nTrain feature std (first 5):")
print(X_train_proc.std(axis=0)[:5])
```

```python
from flax import linen as nn
import jax
import jax.numpy as jnp

class NPE(nn.Module):

    hidden_dim: int = 1024
    n_components: int = 2
    n_params: int = 5

    @nn.compact
    def __call__(self, x):

        # =====================================================
        # Shared trunk
        # =====================================================

        x = nn.leaky_relu(
            nn.Dense(self.hidden_dim)(x),
            negative_slope=0.2
        )

        x = nn.leaky_relu(
            nn.Dense(self.hidden_dim)(x),
            negative_slope=0.2
        )

        x = nn.leaky_relu(
            nn.Dense(self.hidden_dim)(x),
            negative_slope=0.2
        )

        # =====================================================
        # Means
        # =====================================================

        means = nn.Dense(
            self.n_components * self.n_params
        )(x)

        means = means.reshape(
            -1,
            self.n_components,
            self.n_params
        )

        # =====================================================
        # Cholesky params
        # =====================================================

        n_tril = (
            self.n_params *
            (self.n_params + 1)
        ) // 2

        chol_params = nn.Dense(
            self.n_components * n_tril
        )(x)

        chol_params = chol_params.reshape(
            -1,
            self.n_components,
            n_tril
        )

        # =====================================================
        # Mixture weights
        # =====================================================

        logits = nn.Dense(
            self.n_components
        )(x)

        weights = jax.nn.softmax(
            logits,
            axis=-1
        )

        # =====================================================
        # Build covariance matrices
        # =====================================================

        batch_size = x.shape[0]

        L = jnp.zeros(
            (
                batch_size,
                self.n_components,
                self.n_params,
                self.n_params
            )
        )

        tril_idx = jnp.tril_indices(
            self.n_params
        )

        L = L.at[
            :,
            :,
            tril_idx[0],
            tril_idx[1]
        ].set(chol_params)

        diag_idx = jnp.arange(
            self.n_params
        )

        diag_vals = (
            jax.nn.softplus(
                L[:, :, diag_idx, diag_idx]
            )
            + 5e-5
        )

        L = L.at[
            :,
            :,
            diag_idx,
            diag_idx
        ].set(diag_vals)

        covs = jnp.matmul(
            L,
            jnp.swapaxes(L, -1, -2)
        )

        covs = (
            covs
            + 5e-5 * jnp.eye(self.n_params)
        )

        return {
            "means": means,
            "covs": covs,
            "weights": weights
        }
```

```python
import jax
import jax.numpy as jnp


def gaussian_nll(
    y_true,
    means,
    covs,
    weights
):

    """
    y_true:
        (batch, 5)

    means:
        (batch, 2, 5)

    covs:
        (batch, 2, 5, 5)

    weights:
        (batch, 2)
    """

    n_components = means.shape[1]

    log_probs = []

    for k in range(n_components):

        mu = means[:, k]

        C = covs[:, k]

        # Cholesky decomposition

        L = jnp.linalg.cholesky(C)

        # Difference vector

        diff = y_true - mu

        # Mahalanobis term

        solve = jax.scipy.linalg.solve_triangular(
            L,
            diff[..., None],
            lower=True
        )

        maha = jnp.sum(
            solve**2,
            axis=(1, 2)
        )

        # Log determinant

        logdet = 2.0 * jnp.sum(
            jnp.log(
                jnp.diagonal(
                    L,
                    axis1=-2,
                    axis2=-1
                )
            ),
            axis=1
        )

        # Gaussian log-probability

        d = y_true.shape[-1]

        log_prob = -0.5 * (
            maha
            + logdet
            + d * jnp.log(2.0 * jnp.pi)
        )

        log_probs.append(log_prob)

    # Shape = (batch, n_components)

    log_probs = jnp.stack(
        log_probs,
        axis=1
    )

    # Learned mixture weights

    log_weights = jnp.log(
        weights + 1e-12
    )

    # Mixture log probability

    total_log_prob = (
        jax.scipy.special.logsumexp(
            log_probs + log_weights,
            axis=1
        )
    )

    return -jnp.mean(total_log_prob)
```

```python
import jax
import jax.numpy as jnp

# Initialize model

model = NPE()

rng = jax.random.PRNGKey(42)

# Dummy input

dummy_x = jnp.ones((4, 64))

# Initialize parameters

params = model.init(
    rng,
    dummy_x
)

# Forward pass

outputs = model.apply(
    params,
    dummy_x
)

print("Means shape:")
print(outputs["means"].shape)

print("\nCovariance shape:")
print(outputs["covs"].shape)

print("\nWeights shape:")
print(outputs["weights"].shape)

print("\nWeights:")
print(outputs["weights"])
```

```python
import optax

# Stable optimizer

optimizer = optax.chain(

    # Gradient clipping

    optax.clip_by_global_norm(1.0),

    # RMSprop

    optax.rmsprop(
        learning_rate=5e-4
    )
)

# Initialize optimizer state

opt_state = optimizer.init(params)

print("Stable optimizer initialized.")
```

```python
import jax
import jax.numpy as jnp
import optax

@jax.jit
def train_step(
    params,
    opt_state,
    x_batch,
    y_batch
):

    def loss_fn(params):

        outputs = model.apply(
            params,
            x_batch
        )

        loss = gaussian_nll(
            y_batch,
            outputs["means"],
            outputs["covs"],
            outputs["weights"]
        )

        return loss

    loss, grads = jax.value_and_grad(
        loss_fn
    )(params)

    updates, opt_state = optimizer.update(
        grads,
        opt_state,
        params
    )

    params = optax.apply_updates(
        params,
        updates
    )

    return params, opt_state, loss
```

```python
import numpy as np
import jax
import jax.numpy as jnp

# ============================================================
# SKA noise
# ============================================================

def add_ska_noise_catalog(
    X_clean,
    sigma_total
):

    X2d = X_clean.reshape(-1, 8, 8)

    noise = np.random.normal(
        loc=0.0,
        scale=sigma_total,
        size=X2d.shape
    )

    X_noisy = X2d + noise

    X_noisy = np.maximum(
        X_noisy,
        1e-6
    )

    return X_noisy.reshape(-1, 64)


# ============================================================
# Fixed noisy validation set
# ============================================================

X_val_noisy = add_ska_noise_catalog(
    X_val,
    sigma_train
)

X_val_proc_noisy = preprocess_X(
    X_val_noisy,
    X_mean,
    X_std
)

X_val_jax_noisy = jnp.array(
    X_val_proc_noisy
)


# ============================================================
# Training configuration
# ============================================================

batch_size = 32
n_epochs = 400

train_losses = []
val_losses = []

best_val_loss = np.inf


# ============================================================
# Validation loss
# ============================================================

@jax.jit
def eval_loss(
    params,
    x,
    y
):

    outputs = model.apply(
        params,
        x
    )

    return gaussian_nll(
        y,
        outputs["means"],
        outputs["covs"],
        outputs["weights"]
    )


# ============================================================
# Training loop
# ============================================================

n_train = X_train.shape[0]

for epoch in range(n_epochs):

    # ========================================================
    # Fresh SKA realization each epoch
    # ========================================================

    X_train_noisy = add_ska_noise_catalog(
        X_train,
        sigma_train
    )

    X_train_proc_epoch = preprocess_X(
        X_train_noisy,
        X_mean,
        X_std
    )

    X_train_epoch_jax = jnp.array(
        X_train_proc_epoch
    )

    # ========================================================
    # Shuffle
    # ========================================================

    perm = np.random.permutation(
        n_train
    )

    X_train_shuff = X_train_epoch_jax[perm]

    Theta_train_shuff = Theta_train_jax[perm]

    epoch_loss = 0.0
    n_batches = 0

    # ========================================================
    # Mini-batches
    # ========================================================

    for i in range(
        0,
        n_train,
        batch_size
    ):

        x_batch = X_train_shuff[
            i:i+batch_size
        ]

        y_batch = Theta_train_shuff[
            i:i+batch_size
        ]

        if x_batch.shape[0] < batch_size:
            continue

        params, opt_state, loss = train_step(
            params,
            opt_state,
            x_batch,
            y_batch
        )

        if jnp.isnan(loss):

            raise RuntimeError(
                f"NaN loss detected at epoch {epoch+1}"
            )

        params = params
        opt_state = opt_state

        epoch_loss += float(loss)

        n_batches += 1

    # ========================================================
    # Epoch averages
    # ========================================================

    epoch_loss /= n_batches

    val_loss = eval_loss(
        params,
        X_val_jax_noisy,
        Theta_val_jax
    )

    val_loss = float(val_loss)

    # ========================================================
    # Save best model
    # ========================================================

    if val_loss < best_val_loss:

        best_val_loss = val_loss

        best_params = params

    train_losses.append(
        epoch_loss
    )

    val_losses.append(
        val_loss
    )

    print(
        f"Epoch {epoch+1:03d} | "
        f"Train Loss: {epoch_loss:.4f} | "
        f"Val Loss: {val_loss:.4f}"
    )

# ============================================================
# Finished
# ============================================================

print("\nTraining complete.")

print(
    f"Best validation loss = "
    f"{best_val_loss:.6f}"
)
```

```python
import numpy as np
import jax.numpy as jnp

# ============================================================
# Generate ONE noisy SKA realization of test set
# ============================================================

X_test_noisy = add_ska_noise_catalog(
    X_test,
    sigma_train
)

X_test_proc_noisy = preprocess_X(
    X_test_noisy,
    X_mean,
    X_std
)

X_test_noisy_jax = jnp.array(
    X_test_proc_noisy
)

# ============================================================
# Coverage test
# ============================================================

n_examples = min(
    200,
    len(X_test)
)

n_samples = 5000

inside68 = np.zeros(5)
inside95 = np.zeros(5)

all_weights = []

# ============================================================
# Loop over test examples
# ============================================================

for idx in range(n_examples):

    x = X_test_noisy_jax[idx:idx+1]

    outputs = model.apply(
        best_params,
        x
    )

    means = np.array(
        outputs["means"][0]
    )

    covs = np.array(
        outputs["covs"][0]
    )

    weights = np.array(
        outputs["weights"][0]
    )

    all_weights.append(
        weights
    )

    # ========================================================
    # Draw posterior samples
    # ========================================================

    samples = np.zeros(
        (n_samples, 5)
    )

    components = np.random.choice(
        len(weights),
        size=n_samples,
        p=weights
    )

    for j, comp in enumerate(components):

        samples[j] = np.random.multivariate_normal(
            means[comp],
            covs[comp]
        )

    # ========================================================
    # True normalized parameters
    # ========================================================

    theta_true = np.array(
        Theta_test[idx]
    )

    # ========================================================
    # Coverage
    # ========================================================

    for p in range(5):

        low68, high68 = np.percentile(
            samples[:, p],
            [16, 84]
        )

        low95, high95 = np.percentile(
            samples[:, p],
            [2.5, 97.5]
        )

        if low68 <= theta_true[p] <= high68:
            inside68[p] += 1

        if low95 <= theta_true[p] <= high95:
            inside95[p] += 1

# ============================================================
# Final coverage
# ============================================================

inside68 /= n_examples
inside95 /= n_examples

names = [
    "fX",
    "tau",
    "rhs",
    "logMmin",
    "fesc"
]

print("\n===================================")
print("COVERAGE RESULTS")
print("===================================\n")

for i, name in enumerate(names):

    print(
        f"{name:10s}"
        f" 68% = {inside68[i]:.3f}"
        f"   95% = {inside95[i]:.3f}"
    )

# ============================================================
# Mixture diagnostics
# ============================================================

all_weights = np.array(
    all_weights
)

print("\n===================================")
print("MIXTURE WEIGHT DIAGNOSTICS")
print("===================================\n")

print("Mean weights:")
print(
    all_weights.mean(axis=0)
)

print("\nMin weights:")
print(
    all_weights.min(axis=0)
)

print("\nMax weights:")
print(
    all_weights.max(axis=0)
)

# ============================================================
# Calibration summary
# ============================================================

print("\n===================================")
print("CALIBRATION SUMMARY")
print("===================================\n")

print(
    "Ideal 68% coverage = 0.68"
)

print(
    "Ideal 95% coverage = 0.95"
)

print(
    "\nMean 68% coverage =",
    inside68.mean()
)

print(
    "Mean 95% coverage =",
    inside95.mean()
)
```

```python
import numpy as np
import matplotlib.pyplot as plt
import jax.numpy as jnp

# ============================================================
# One noisy SKA realization
# ============================================================

X_test_noisy = add_ska_noise_catalog(
    X_test,
    sigma_train
)

X_test_proc_noisy = preprocess_X(
    X_test_noisy,
    X_mean,
    X_std
)

X_test_noisy_jax = jnp.array(
    X_test_proc_noisy
)

# ============================================================
# Storage
# ============================================================

n_samples = 5000
n_test = len(X_test)

true_all = []
median_all = []
sigma_all = []

# ============================================================
# Loop over test set
# ============================================================

for idx in range(n_test):

    x = X_test_noisy_jax[idx:idx+1]

    outputs = model.apply(
        best_params,
        x
    )

    means = np.array(
        outputs["means"][0]
    )

    covs = np.array(
        outputs["covs"][0]
    )

    weights = np.array(
        outputs["weights"][0]
    )

    # --------------------------------------------------------
    # Posterior samples
    # --------------------------------------------------------

    samples = np.zeros(
        (n_samples, 5)
    )

    comps = np.random.choice(
        len(weights),
        size=n_samples,
        p=weights
    )

    for j, comp in enumerate(comps):

        samples[j] = np.random.multivariate_normal(
            means[comp],
            covs[comp]
        )

    # --------------------------------------------------------
    # True normalized theta
    # --------------------------------------------------------

    theta_true = np.array(
        Theta_test[idx]
    )

    # --------------------------------------------------------
    # Median and sigma
    # --------------------------------------------------------

    medians = []
    sigmas = []

    for p in range(5):

        q16,q50,q84 = np.percentile(
            samples[:,p],
            [16,50,84]
        )

        medians.append(q50)

        sigmas.append(
            0.5*(q84-q16)
        )

    true_all.append(theta_true)

    median_all.append(medians)

    sigma_all.append(sigmas)

# ============================================================
# Convert arrays
# ============================================================

true_all = np.array(
    true_all
)

median_all = np.array(
    median_all
)

sigma_all = np.array(
    sigma_all
)

# ============================================================
# Recovery plots
# ============================================================

names = [
    "fX",
    "tau",
    "rhs",
    "logMmin",
    "fesc"
]

fig, axes = plt.subplots(
    2,
    3,
    figsize=(12,8)
)

axes = axes.flatten()

for p in range(5):

    ax = axes[p]

    ax.scatter(
        true_all[:,p],
        median_all[:,p],
        s=10,
        alpha=0.5
    )

    xmin = min(
        true_all[:,p].min(),
        median_all[:,p].min()
    )

    xmax = max(
        true_all[:,p].max(),
        median_all[:,p].max()
    )

    ax.plot(
        [xmin,xmax],
        [xmin,xmax],
        "k--"
    )

    ax.set_title(
        names[p]
    )

    ax.set_xlabel(
        "True"
    )

    ax.set_ylabel(
        "Posterior Median"
    )

axes[5].axis("off")

plt.tight_layout()
plt.show()

# ============================================================
# z-score test
# ============================================================

z = (
    median_all
    -
    true_all
) / sigma_all

print("\n===================================")
print("Z-SCORE DIAGNOSTICS")
print("===================================\n")

for p,name in enumerate(names):

    mean_z = np.mean(
        z[:,p]
    )

    std_z = np.std(
        z[:,p]
    )

    print(
        f"{name:10s}"
        f" mean(z)={mean_z:.3f}"
        f" std(z)={std_z:.3f}"
    )

# ============================================================
# Histogram
# ============================================================

plt.figure(
    figsize=(8,5)
)

for p,name in enumerate(names):

    plt.hist(
        z[:,p],
        bins=30,
        histtype="step",
        density=True,
        label=name
    )

plt.axvline(
    0,
    ls="--"
)

plt.xlabel(
    r"$z=(\theta_{\rm med}-\theta_{\rm true})/\sigma$"
)

plt.ylabel(
    "Density"
)

plt.legend()

plt.tight_layout()

plt.show()
```
